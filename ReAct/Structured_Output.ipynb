```python
from langchain_google_genai import ChatGoogleGenerativeAI
from pydantic import BaseModel, Field
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
llm.invoke("What is the capital of France?")
```

### Adding output format in the question only!

```python
llm.invoke("What is the capital of France? Generate the output with format:: " 'Query: <the query>' ' Answer: <the answer>')
```

### Using Pydantic library

```python
class llm_schema(BaseModel):
    query: str = Field(description="The query asked to the LLM")
    answer: str = Field(description="The answer returned by the LLM")
    total_tokens: int = Field(description="The total number of tokens used by the LLM")
```

```python
obj = llm_schema(query="What is the capital of France?", answer="The capital of France is Paris.", total_tokens=10)
obj
```

### Provide the feature for validation also

```python
obj = llm_schema(query="What is the capital of France?", answer="The capital of France is Paris.", total_tokens="Total tokens are: 10")
obj
```

#### Integrate with LLM

```python
str_output = llm.with_structured_output(schema=llm_schema)

response = str_output.invoke("What is the capital of France?")
response
```

```python
final = response.model_dump()  ## convert into dictionary
final
```
