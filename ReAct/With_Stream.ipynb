```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
load_dotenv()
```

```python
llm_model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

#### Tools

```python
## Tool 1: Search tool
from langchain_community.tools import DuckDuckGoSearchRun
search = DuckDuckGoSearchRun(description = "This is a tool to search the web")
search.invoke("Who is Sunder?")

## Tool 2: Wikipedia search tool
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
wikipedia = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper(), description = "This is a tool to search wikipedia.")
wikipedia.run("Sam Altman?")

## Tool 3: Custom tool
from langchain.tools import tool

@tool
def enterprise_tool(query: str) -> str:
    """
    This is a tool to send emails to employees.
    """
    return "Email sent"
```

#### Toolkit

```python
toolkit = [search, wikipedia, enterprise_tool]
toolkit
```

### ReAct Agent

```python
from langchain.agents import create_agent

my_agent = create_agent(llm_model, tools = toolkit)
my_agent
```

### Agent invoke With Stream

```python
query = "Give me the latest news?"


events = my_agent.stream(
    { "messages" : [ ("user", query) ] },
    stream_mode = "values"
)

for event in events:
    event["messages"][-1].pretty_print()
```
