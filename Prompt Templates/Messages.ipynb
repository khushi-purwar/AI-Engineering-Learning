```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from langchain.chat_models import init_chat_model

import os
import load_dotenv
# this will load all the environment variables from .env file and will
# make available to the os.environ dictionary
load_dotenv()

llm_call = init_chat_model(
    model="gemini-2.5-flash",
    temperature=0
)

my_messages = [
    SystemMessage(content="You are a gen-z assistant who always answers in a fun way"),
    HumanMessage(content="Bro, tell me a fun fact")
]

llm_call.invoke(my_messages).content
```
