```python
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

### Define LLM

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
llm.invoke("Write a short poem about the beauty of nature.")
```

```python
message = [
    {"role": "system", "content": "You are a genZ poet."},
    {"role": "user", "content": "Write a short poem about the beauty of nature."}
]

response = llm.invoke(message)
response
```

```python
response.content
```

### Use PromptTemplate

```python
from langchain_core.prompts import PromptTemplate
```

```python
my_prompt = PromptTemplate.from_template("You are a genZ poet. Write a short poem about the {topic}.")
my_prompt
```

```python
filled_prompt = my_prompt.format(topic="beauty of nature") ### fill with a topic
filled_prompt
# a simple string prompt is created.
```

### Use ChatPromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
my_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a genZ poet."),
        ("user", "Write a short poem about the beauty of nature.")
    ]
)

my_template ## no need to write role, and content. Just mention the role and content in a tuple. The first element is the role and the second element is the content.
```

### Now make it dynamic, need to write poem on different topics

```python
my_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a genZ poet."),
        ("user", "Write a short poem about the {topic}")
    ]
)
my_template
```

```python
final_message  = my_template.format_messages(topic = "beauty of nature")
final_message
```

```python
llm.invoke(final_message)
```
