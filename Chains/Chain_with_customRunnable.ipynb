#### Import Libraries

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

#### Task 1: Prompt

```python
my_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a genZ poet."),
        ("user", "Write a short poem about the {topic}")
    ]
)
my_template
```

#### Task 2 - LLM

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

##### Task 3 - Parser

```python
str_parser = StrOutputParser()
```

#### Task 4 - Custom Runnable Functions

```python
from langchain_core.runnables import RunnableLambda

def dictionary_maker(text: str) -> dict:
    return {"text" : text}

dictionary_maker_runnable = RunnableLambda(dictionary_maker)
```

#### Task 5 - Template for Post

```python
prompt_post = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a social media post generator."),
        ("user", "Create a post for the following text for LinkedIn: {text}")
    ]
)
```

#### Task 6 - Generate the Post (LLM)

```python
llm_generate = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

#### Task 7 - Parser

```python
str_parser_last = StrOutputParser()
```

```python
### Chain

chain = my_template | llm | str_parser | dictionary_maker_runnable | prompt_post | llm_generate | str_parser_last
chain.invoke("What is the capital of France ?")
```
